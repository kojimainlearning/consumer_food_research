#### All Parameters

```python
import mne
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
%matplotlib qt
```

```python
def load_raw_edf(file_path, verbose=True):
    file_path = file_path if file_path != None or "" else "ensemble_de_donnees/eau/lkq_eau.edf"
    raw_eeg = mne.io.read_raw_edf(file_path, preload=True, verbose=verbose)

    eeg_channels = ["AF3", "F7", "F3", "FC5", "T7", "P7", "O1", "O2", "P8", "T8", "FC6", "F4", "F8", "AF4"]
    raw_eeg.pick(eeg_channels, verbose=verbose)

    montage = mne.channels.make_standard_montage("standard_1020")
    raw_eeg.set_montage(montage, verbose=verbose)

    return raw_eeg
```

### Preprocessing

##### Bandpass Filter

```python
def bandpass_filter(raw_eeg, l_freq, h_freq, verbose=True):
    raw_filtered = raw_eeg.copy().filter(
        l_freq=l_freq, 
        h_freq=h_freq, 
        fir_design="firwin", 
        verbose=verbose)
    return raw_filtered
```

##### Independent Component Analysis (ICA)

```python
def noise_removal_using_ica(raw_filtered, ica_components, verbose=True):
    ica = mne.preprocessing.ICA(n_components=ica_components, random_state=42, verbose=verbose, max_iter=1000)
    ica.fit(raw_filtered)

    eog_indices, eog_scores = ica.find_bads_eog(raw_filtered, ch_name=['F7', 'F8'], verbose=verbose)
    print(f"Found EOG artifacts in components: {eog_indices}")
    ica.exclude = eog_indices

    muscle_indices, muscle_scores = ica.find_bads_muscle(raw_filtered, verbose=verbose)
    print(f"Found muscle artifacts in components: {muscle_indices}")
    ica.exclude = muscle_indices

    raw_clean = ica.apply(raw_filtered.copy(), verbose=verbose)

    return raw_clean
```

##### Epoching and Artifact Rejection

```python
def epoch_and_interpolate_bads(raw_clean, epoch_length, reject_criteria, verbose=True):
    events = mne.make_fixed_length_events(raw_clean, duration=epoch_length, overlap=epoch_length * 0.5)
    epochs = mne.Epochs(raw_clean, events, tmin=0, tmax=epoch_length, baseline=None, preload=True, verbose=verbose)

    clean_epochs = []

    for i in range(len(epochs)):
        epoch = epochs[i].get_data()[0]
        ptp_amplitude = np.ptp(epoch, axis=1)  # Peak-to-peak amplitude for each channel
        
        bad_ch_idx = np.where(ptp_amplitude > reject_criteria)[0]
        bad_channels = [epochs.ch_names[idx] for idx in bad_ch_idx]
        
        if len(bad_channels) == 0:
            clean_epochs.append(epochs[i].copy())
            
        elif len(bad_channels) <= 3:
            if verbose:
                print(f"Epoch {i} has bad channels: {bad_channels}")
                
            temp_epoch = epochs[i].copy()
            temp_epoch.info["bads"] = bad_channels
            temp_epoch.interpolate_bads(reset_bads=True, verbose=verbose) # Interpolate bad channels with good channels
            
            clean_epochs.append(temp_epoch)
        
        else:
            if verbose:
                print(f"Epoch {i} has too many bad channels: {bad_channels}. Discarding this epoch.")
        
        
    if verbose:
        if len(clean_epochs) > 0:
            print(f"Number of original epochs: {len(epochs)}")
            print(f"Number of remaining clean epochs: {len(clean_epochs)}")
        else:
            print("No clean epochs remaining after artifact rejection.")

    
    if len(clean_epochs) == 0:
        return None    
        
    return mne.concatenate_epochs(clean_epochs, verbose=verbose)
```

#### Power Spectral Density with Welch's Method using FFT

```python
def extract_metrics(clean_epochs, verbose=True):
    spectrum = clean_epochs.compute_psd(method="welch", fmin=1, fmax=45, n_fft=256, n_overlap=128, verbose=verbose)
    psds, freqs = spectrum.get_data(return_freqs=True)

    def get_band_power(psds, freqs, band):
        # Check if the specified frequency band is valid
        if len(band) != 2 or band[0] >= band[1]:
            raise ValueError(f"Invalid frequency band {band}. It should be a tuple of (low_freq, high_freq) with low_freq < high_freq.")
        elif band[0] < freqs[0] or band[1] > freqs[-1]:
            raise ValueError(f"Frequency band {band} is out of range. Valid range is {freqs[0]}-{freqs[-1]} Hz.")
        
        # Extract the average power in the specified frequency band
        band_idx = np.logical_and(freqs >= band[0], freqs <= band[1])
        band_power = np.mean(psds[:, :, band_idx], axis=2)
        return band_power

    # Extract power from all epochs
    theta_band = (4, 8)
    alpha_band = (8, 13)
    beta_band = (13, 30)
    
    theta_power = get_band_power(psds, freqs, theta_band)
    alpha_power = get_band_power(psds, freqs, alpha_band)
    beta_power = get_band_power(psds, freqs, beta_band)

    ch_names = clean_epochs.ch_names

    # Calculate Metrics using Different Frequency Bands
    # Channels for happiness
    f3_idx = ch_names.index('F3')
    f4_idx = ch_names.index('F4')
    
    # Channels for calmness
    calmness_channels = ['O1', 'O2']
    calmness_idx = [ch_names.index(ch) for ch in calmness_channels]
    
    
    faa_per_epoch = np.log(alpha_power[:, f4_idx] + 1e-12) - np.log(alpha_power[:, f3_idx] + 1e-12)
    
    avg_alpha_calm = np.mean(alpha_power[:, calmness_idx], axis=1)
    avg_beta_calm = np.mean(beta_power[:, calmness_idx], axis=1)
    calmness_per_epoch = avg_alpha_calm / (avg_beta_calm + 1e-12)
    
    final_faa = np.mean(faa_per_epoch)
    final_calmness = np.mean(calmness_per_epoch)

    if verbose:
        print(f"Total Clean Epochs Used : {len(clean_epochs)}")
        print(f"Happiness Score         : {final_faa:.4f}")
        print(f"Calmness Index          : {final_calmness:.4f}")

    return final_faa, final_calmness
```

```python
VERBOSE_ON = True
LOW_CUTOFF = 1.0
HIGH_CUTOFF = 45.0
ICA_COMPONENTS = 14

file_path = "ensemble_de_donnees/eau/lkq_eau.edf"
raw_eeg = load_raw_edf(file_path)

raw_before = raw_eeg.copy().crop(tmin=0, tmax=6)
raw_after = raw_eeg.copy().crop(tmin=7, tmax=20)

EPOCH_LENGTH_BEFORE = 2
REJECT_CRITERIA_BEFORE = 200e-6

EPOCH_LENGTH_AFTER = 4
REJECT_CRITERIA_AFTER = 120e-6
```

```python
raw_filtered_before = bandpass_filter(raw_before, LOW_CUTOFF, HIGH_CUTOFF, VERBOSE_ON)
raw_clean_before = noise_removal_using_ica(raw_filtered_before, ICA_COMPONENTS, VERBOSE_ON)
clean_epochs_before = epoch_and_interpolate_bads(raw_clean_before, EPOCH_LENGTH_BEFORE, REJECT_CRITERIA_BEFORE, VERBOSE_ON)
faa_before, calmness_before = extract_metrics(clean_epochs_before, VERBOSE_ON)
```

```python
raw_filtered_after = bandpass_filter(raw_after, LOW_CUTOFF, HIGH_CUTOFF, VERBOSE_ON)
raw_clean_after = noise_removal_using_ica(raw_filtered_after, ICA_COMPONENTS, VERBOSE_ON)
clean_epochs_after = epoch_and_interpolate_bads(raw_clean_after, EPOCH_LENGTH_AFTER, REJECT_CRITERIA_AFTER, VERBOSE_ON)
faa_after, calmness_after = extract_metrics(clean_epochs_after, VERBOSE_ON)
```
